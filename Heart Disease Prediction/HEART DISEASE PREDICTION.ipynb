## Data Preprocessing

```python
data.drop(columns="patient_id",inplace = True)
```

```python
# Create New features
data["age_oldpeak"] = data["age"]*data["oldpeak_eq_st_depression"]
```

```python
data.head()
```

## Conversion of catagorical variable into a numerical variable

```python
data["thal"] = data["thal"].map({"fixed_defect":0,"reversible_defect":1,"normal":2})
```

```python
# Insights
 # Use a manual mapping to convert the categorical data into numerical data
```

## Handle outliers (Skewness)

```python
continuous_skew = ["serum_cholesterol_mg_per_dl","oldpeak_eq_st_depression","max_heart_rate_achieved","age","oldpeak_eq_st_depression","age_oldpeak"]
```

```python
for col in continuous_skew:
    print(col , data[col].skew())
```

```python
for col in continuous_skew:
    if data[col].skew() > 1:
        data[col] = np.log1p(data[col])
```

```python
# Insights
 # We don't handle the outliers because they  are a valid information so we use log transform to reduce the skewness
```

## Split the data

```python
x=data.drop("heart_disease_present",axis=1)
y=data.heart_disease_present
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,random_state = 42)
```

```python
x_train_scaled = x_train.copy()
x_test_scaled = x_test.copy()
```

## Scale the Continuous data

```python
from sklearn.preprocessing import StandardScaler
scale = StandardScaler()
x_train_scaled[continuous_skew] = scale.fit_transform(x_train[continuous_skew])
x_test_scaled[continuous_skew] = scale.transform(x_test[continuous_skew])
```

## Model Building

```python
from sklearn.linear_model import LogisticRegression
model = LogisticRegression(random_state=30)
model.fit(x_train_scaled,y_train)
y_pred = model.predict (x_test_scaled)
y_pred
```

```python
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
print(accuracy_score(y_test,y_pred)*100)
```

```python
print(classification_report(y_test,y_pred))
```

```python
print(confusion_matrix(y_test,y_pred))
```

## Cross Validation

```python
from sklearn.model_selection import cross_val_score
x_scale = scale.fit_transform(x)
score_recall = cross_val_score(
    model,
    x_scale,
    y,
    cv=5,
    scoring='recall',
)
print("Recall scores:", score_recall)
print("Mean Recall:", score_recall.mean())
```

```python
# Insights
 # Logistic Regression achieved 84% accuracy on the test set with a strong and balanced F1-score of 0.84.
 # The model shows 73% recall for heart disease cases, meaning it correctly identifies most positive patients.
 # 5-Fold Cross Validation produced an average recall of 75%, which is close to the test recall.
 # This small gap indicates the model is stable, generalizes well, and is not overfitting for the 180-row dataset
```

## Decision Tree

```python
from sklearn.tree import DecisionTreeClassifier
dt_model = DecisionTreeClassifier(random_state = 42)
dt_model.fit(x_train,y_train)
y_predict = dt_model.predict(x_test)
y_predict
```

```python
print(accuracy_score(y_test,y_predict)*100)
```

## Hyperparameter Tuning

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import GridSearchCV

param_grid = {
    'max_depth': [2, 3, 4, 5],
    'min_samples_split': [5, 10, 15],
    'min_samples_leaf': [2, 4, 6],
    'criterion': ['gini', 'entropy']
}

dt = DecisionTreeClassifier(random_state=42)
grid = GridSearchCV(
    dt,
    param_grid,
    cv=5,
    scoring='accuracy',
    n_jobs=-1
)
grid.fit(x, y)
print("Best Parameters:", grid.best_params_)
print("Best CV Recall:", grid.best_score_)
```

```python
dt_model1 = DecisionTreeClassifier(criterion= 'entropy', max_depth= 4, min_samples_leaf= 4, min_samples_split= 10,random_state = 42)
dt_model1.fit(x_train,y_train)
y_predict1 = dt_model1.predict(x_test)
y_predict1
```

```python
print(accuracy_score(y_test,y_predict1)*100)
```

```python
print(classification_report(y_test,y_predict1))
```

```python
# Insights
 # The default Decision Tree achieved around 75% accuracy with 65% recall, showing moderate heart disease detection but some instability due to the small dataset.
 # After hyperparameter tuning, performance improved to 80% accuracy with 69% recall, indicating better balance and reduced overfitting.
 # Tuning helped control model complexity and improved generalization.
 # However, despite improvement, the tuned Decision Tree still performed slightly below Logistic Regression in overall stability and recall.
```
